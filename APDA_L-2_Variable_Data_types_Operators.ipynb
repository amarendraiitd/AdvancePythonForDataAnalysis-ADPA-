## What is Escape sequence character?

An escape sequence is a special character combination starting with \ used to represent formatting or special characters inside strings.


| Escape Sequence | Meaning         | Example            | Output            |
| --------------- | --------------- | ------------------ | ----------------- |
| `\n`            | New line        | `"Hello\nWorld"`   | Hello<br>World    |
| `\t`            | Tab (space)     | `"A\tB"`           | A  B              |
| `\\`            | Backslash       | `"C:\\Users"`      | C:\Users          |
| `\'`            | Single quote    | `'It\'s good'`     | It's good         |
| `\"`            | Double quote    | `"He said \"Hi\""` | He said "Hi"      |
| `\r`            | Carriage return | `"Hello\rHi"`      | Hi                |
| `\b`            | Backspace       | `"Helloo\b"`       | Hello             |
| `\f`            | Form feed       | `"Hello\fWorld"`   | Page break (rare) |
| `\v`            | Vertical tab    | `"A\vB"`           | Vertical spacing  |

## Numbers in python
- Python has three primary built-in numeric data types for handling mathematical values: integers, floating-point numbers, and complex numbers. 

1. Integers (int):

    - integers are whole numbers (positive, negative, or zero) that belong to the int class and have arbitrary precision, meaning their size is limited only by your computer's memory.  
    - int Class: All integers in Python are objects of the built-in int class.
    - Examples: 10, -5, 0

2. Floating-Point Numbers (float): 
    - These represent real numbers with a decimal point and are accurate up to approximately 15 decimal places. 
    - They can also be written in scientific notation using e or E (e.g., 1.5e6 equals 1,500,000).
    - Examples: 3.14, -0.5, 2.0

3. Complex Numbers (complex):
    - These are numbers with a real and an imaginary part, written in the form a + bj, where j denotes the imaginary unit.
    - Examples: 3 + 4j, 1j

## Variable Assignment in python?

1. Object: 
    - In Python, Everything is an object.
    - An object has:
        1. Identity → Unique ID (memory address)
        2. Type → int, float, str, list, etc.
        3. Value → actual data
    - An object is a piece of memory that holds data (like the number 257 or the string "Hello").
    
2. Variable:
    - Common misconception: “Variable is a box that stores a value”.
    - Python reality: A variable is just a name (label/interpreter) that points to an object. 

    - x = 10
        - x → name
        - = → assignment
        - 10 → object

    - Python has names, not variables.
    - Python is a dynamically typed language: Names can point to any type of object.


- How Variable / Object / Value Are Stored in Memory:
    - x = 2
    - What Python does internally:
        - Creates integer object 2
        - Stores it in heap memory (Heap memory is a region of memory where objects are dynamically created and stored at runtime.)
        - Creates name x
        - Binds x → address of object 2 
        - x ──► 2 
        - print(id(x))  # memory identity
        - Names live in namespace
        - Objects live in heap

- Interning?
    - Python tries to save memory by reusing common objects. This is called Interning.
    - Integer interning: Python pre-creates integer objects from -5 to 256 when the program starts.These stay in memory the whole time.
    - Any time you use x=2, Python points your variable to the already existing 2 in the pool.
    
    - Multiple variables can point to the same object (and thus the same memory address).
    - If you use a much larger number, they might point to different addresses depending on whether you are running the code in a script or an interactive shell(REPL).



```python
x = 2
y = x

print(id(x))
print(id(y))

print(x is y)
print(x == y)

# x ──► 2
# y ──► 2
# same object, with two different names
```

- Global Variables:
    - variables that are created outside of a function. can be used by everyone inside and outside of function.
    - normally, when you create a variable inside a function, that is local variable and can only be used inside that function.
    - to create a global var inside a fun, use the 'global' keyword.

```python
# Ex create a var outside of a fun and use it inside fun,
x = "good"

def myfun():
    print("you are "+ x)

myfun()
```

```python
# Ex create a var inside a fun with the same name as the global var
x = "good"

def myfun():
    x = "better"
    print("you are "+ x)

myfun()

print("you are "+ x)
```

```python
# Ex, in the above problem. to create a global var inside a fun, use 'global' keyword

x = "good"

def myfun():
    global x
    x = "better"
    print("you are "+ x)

myfun()

print("you are "+ x)
```

| -------- | -------------- | ------------------------ | -------- | ------ |
