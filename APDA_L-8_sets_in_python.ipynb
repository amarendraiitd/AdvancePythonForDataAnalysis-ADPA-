## What are Sets?

- Sets are an unordered collection data type that is iterable, mutable, and has no duplicate elements.
- Python's set class represents the mathematical notion of a set.
- This is a key difference from other collection types like lists and tuples, which allow duplicate elements.

### Characteristics of Sets:

1.  **Unordered**: Elements in a set do not have a specific order. When you print a set, the elements may appear in a different order than they were inserted.
2.  **Mutable**: You can add or remove elements from a set after it has been created.
3.  **No Duplicate Elements**: Sets automatically remove duplicate entries. Each element in a set must be unique.
4.  **Iterable**: You can loop through the elements of a set.
5.  **Heterogeneous**: A set can contain elements of different data types (e.g., numbers, strings, tuples). However, elements themselves must be immutable (e.g., numbers, strings, tuples can be elements, but lists and dictionaries cannot).


### Creating Sets

- Sets can be created using curly braces `{}` or the `set()` constructor.
- An empty set must be created using `set()`, not `{}` (as `{}` creates an empty dictionary).

```python
# Creating an empty set
empty_set = set()

empty_set
```

```python
# Creating a set with elements using curly braces
my_set = {1, 2, 3, 4, 5}

my_set
```

```python
# Creating a set with mixed data types
mixed_set = {1, 'hello', 3.14, (1, 2)}

mixed_set
```

```python
# Creating a set from a list (duplicates are automatically removed)
list_with_duplicates = [1, 2, 2, 3, 4, 4, 5]
set_from_list = set(list_with_duplicates)

set_from_list
```

```python
# Trying to create a set with mutable elements (e.g., a list inside a set) will raise an error

set_with_list = {1, [2, 3]}
print(set_with_list)
```

### Common Set Operations

Sets support a variety of operations to manipulate their elements. These include adding new elements, removing existing ones, and checking for the presence of an element.

#### Adding Elements

-   `add()`: Adds a single element to the set. If the element is already present, the set remains unchanged.
-   `update()`: Adds elements from an iterable (like a list, tuple, or another set) to the current set. Duplicate elements are ignored.

```python
my_set = {1, 2, 3}
print("Original set:", my_set)

# Adding a single element
my_set.add(4)
print("After adding 4:", my_set)

# Adding an existing element (set remains unchanged)
my_set.add(2)
print("After adding 2 again:", my_set)

# Updating with another set
my_set.update({5, 6})
print("After updating with {5, 6}:", my_set)

# Updating with a list (duplicates are handled)
my_set.update([6, 7, 8])
print("After updating with [6, 7, 8]:", my_set)
```

#### Removing Elements

-   `remove(element)`: Removes a specified element from the set. If the element is not found, it raises a `KeyError`.
-   `discard(element)`: Removes a specified element from the set. If the element is not found, the set remains unchanged (no error).
-   `pop()`: Removes and returns an arbitrary element from the set. Since sets are unordered, you cannot predict which element will be removed. Raises `KeyError` if the set is empty.
-   `clear()`: Removes all elements from the set, making it an empty set.

```python
my_set = {10, 20, 30, 40, 50}

# Using remove()
my_set.remove(20)
my_set
```

```python
# Trying to remove a non-existent element with remove() will raise an error

my_set.remove(99)
my_set
```

```python
# Using discard()
my_set.discard(40)
my_set
```

```python
# Discarding a non-existent element (no error)
my_set.discard(99)
my_set
```

```python
# Using pop()
popped_element = my_set.pop()
print("Popped element:", popped_element)
print("Set after pop:", my_set)
```

```python
# Using clear()
my_set.clear()
my_set
```

#### Checking for Elements (Membership Test)

You can check if an element exists in a set using the `in` and `not in` operators. This is very efficient for sets due to their underlying data structure.

```python
my_set = {'apple', 'banana', 'cherry'}

# Check for existence
print('banana' in my_set)
print('grape' in my_set)

# Check for non-existence
print('orange' not in my_set)
print('apple' not in my_set)
```

### Mathematical Set Operations

- Sets are powerful for performing common mathematical set operations.
- These operations are essential when you need to compare or combine sets based on their elements.

#### Union

-   `union()` or `|` operator: Returns a new set containing all unique elements from both sets. It combines elements from both sets without duplicates.

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}

# Using union() method
union_set_method = set1.union(set2)
union_set_method
```

```python
# Using | operator
union_set_operator = set1 | set2
union_set_operator
```

```python
set3 = {'a', 'b'}
set4 = {'b', 'c'}
set3.union(set4)
```

#### Intersection

-   `intersection()` or `&` operator: Returns a new set containing only the elements common to both sets.

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

# Using intersection() method
intersection_set_method = set1.intersection(set2)
print("Intersection (method):", intersection_set_method)

# Using & operator
intersection_set_operator = set1 & set2
print("Intersection (operator):", intersection_set_operator)

set3 = {'apple', 'banana'}
set4 = {'banana', 'cherry'}
print("Intersection of set3 and set4:", set3.intersection(set4))
```

#### Difference

-   `difference()` or `-` operator: Returns a new set containing elements that are in the first set but not in the second set.

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

# Using difference() method
difference_set_method = set1.difference(set2)
print("Elements in set1 but not in set2 (method):", difference_set_method)

# Using - operator
difference_set_operator = set1 - set2
print("Elements in set1 but not in set2 (operator):", difference_set_operator)

print("Elements in set2 but not in set1:", set2 - set1)
```

#### Symmetric Difference

-   `symmetric_difference()` or `^` operator: Returns a new set containing all elements that are in either set, but not in both (i.e., elements unique to each set).

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

# Using symmetric_difference() method
symmetric_diff_method = set1.symmetric_difference(set2)
print("Symmetric Difference (method):", symmetric_diff_method)

# Using ^ operator
symmetric_diff_operator = set1 ^ set2
print("Symmetric Difference (operator):", symmetric_diff_operator)
```

### Set Comparisons

Python sets also provide methods to compare sets, checking for relationships like subset, superset, and disjointness.

#### issubset()

-   `issubset()` or `<=` operator: Returns `True` if all elements of one set are present in another set. `True` if the first set is a subset of the second set.

```python
set_a = {1, 2, 3}
set_b = {1, 2, 3, 4, 5}

print("Is set_a a subset of set_b (method)?", set_a.issubset(set_b))
print("Is set_b a subset of set_a (method)?", set_b.issubset(set_a))

print("Is set_a a subset of set_b (operator <=)?", set_a <= set_b)
print("Is set_b a subset of set_a (operator <=)?", set_b <= set_a)

# Proper subset (strictly smaller, < operator)
print("Is set_a a proper subset of set_b (operator <)?", set_a < set_b)
print("Is set_b a proper subset of set_a (operator <)?", set_b < set_a)
```

#### issuperset()

-   `issuperset()` or `>=` operator: Returns `True` if all elements of one set are present in the first set. `True` if the first set is a superset of the second set.

```python
set_a = {1, 2, 3, 4, 5}
set_b = {1, 2, 3}

print("Is set_a a superset of set_b (method)?", set_a.issuperset(set_b))
print("Is set_b a superset of set_a (method)?", set_b.issuperset(set_a))

print("Is set_a a superset of set_b (operator >=)?", set_a >= set_b)
print("Is set_b a superset of set_a (operator >=)?", set_b >= set_a)

# Proper superset (strictly larger, > operator)
print("Is set_a a proper superset of set_b (operator >)?", set_a > set_b)
print("Is set_b a proper superset of set_a (operator >)?", set_b > set_a)
```

#### isdisjoint()

-   `isdisjoint()`: Returns `True` if two sets have no elements in common (their intersection is an empty set). Otherwise, it returns `False`.

```python
set_x = {1, 2, 3}
set_y = {4, 5, 6}
set_z = {3, 7, 8}

print("Are set_x and set_y disjoint?", set_x.isdisjoint(set_y))
print("Are set_x and set_z disjoint?", set_x.isdisjoint(set_z))
print("Are set_y and set_z disjoint?", set_y.isdisjoint(set_z))
```
