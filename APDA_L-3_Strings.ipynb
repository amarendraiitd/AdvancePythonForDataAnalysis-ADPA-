### String formating: 

- It is the technique of inserting variables and expressions into strings in a readable and efficient way.

- Why string formatting is needed?
    - Without formatting, printing values becomes messy and unreadable.
    
1. .format() method
2. f-strings

```python
name = "Aman"
age = 25
```

```python
print("My name is " + name + " and my age is " + str(age)) # bad pratice
```

```python
# f-Strings (BEST & MOST IMPORTANT) - Introduced in Python 3.6+

name = "Aman"
age = 25

print(f"My name is {name} and my age is {age}")
```

```python
# .format() method - introduced in python 3

name = "Aman"
age = 25

print("My name is {} and my age is {}".format(name, age))
```

```python
name = "Aman"
age = 25

txt = "My name is {} and my age is {}"

print(txt.format(name, age))
```

```python
print("Age: {1}, Name: {0}".format(name, age)) # with index
```

```python
print("Name: {n}, Age: {a}".format(n=name, a=age)) # with named placeholders
```

```python
name = input("Enter name: ")
marks = int(input("Enter marks: "))

print(f"Student {name} scored {marks} marks")
```

    upper() - Converts all characters to uppercase

    lower() - Converts all characters to lowercase

    capitalize() - Capitalizes only the first character.

    title() - Capitalizes the first letter of every word

    swapcase() - Converts uppercase → lowercase and vice versa

    strip() - Removes white spaces from both sides

    lstrip() - Removes white spaces from left side only

    rstrip() - Removes white spaces from right side only (or any trailing character).

```python
text = "  hello python!!!  "

text.upper()
```

```python
text.lower()
```

```python
text = "  hello python!!!  "

text.strip()  # Removing extra spaces 
```

```python
text.lstrip() 
```

```python
text.rstrip()
```

```python
text = "  hello python!!!  "

text.lstrip("!")
```

```python
text.rstrip("!")
```

```python
text2 = "!!!hello python!!!"
```

```python
text2.lstrip("!")
```

```python
text2.rstrip("!")
```

```python
text = "  hello python!!!  "

text.replace("python", "AI")  # Replacing text
```

    split() method:

```python
text2 = "hello python learners"

text2.split()         # Splitting a string into list of substrings based on whitespace
```

```python
text2 = "hello,python,learners"

text2.split(',')      # split based on (,)
```

```python
# Limit the number of splits using maxsplit

log = "2025-12-16 10:00:00 ERROR Server down"

parts = log.split(" ", maxsplit=3)
parts
```
