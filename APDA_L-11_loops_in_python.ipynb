# Loops in python:

- Loops are fundamental constructs in programming that allow you to execute a block of code repeatedly.
- This is incredibly useful for tasks that involve processing collections of data, performing actions a certain number of times, or continuing an operation until a specific condition is met.

## 1. `for` Loop

- The `for` loop is used for iterating over a sequence (that is, a list, tuple, dictionary, set, or string) or other iterable objects.
- It executes a block of code for each item in the sequence.

```python
# Example: Iterating over a list

fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)
```

```python
# Example: Iterating over a string

for char in "Python":
  print(char)
```

```python
# Example: Using range() function
# The range() function returns a sequence of numbers, starting from 0 by default,
# and increments by 1 (by default), and stops before a specified number.

for i in range(5): # This will print numbers from 0 to 4
  print(i)
```

```python
for i in range(2, 6): # This will print numbers from 2 to 5
  print(i)
```

```python
for i in range(0, 10, 2): # This will print even numbers from 0 to 8
  print(i)
```

## 2. `while` Loop

The `while` loop executes a block of code as long as a specified condition is true. It's often used when you don't know in advance how many times the loop will need to run.

```python
# Example: Counting up to 5

i = 1
while i < 6:
  print(i)
  i = i + 1
```

```python
# Example: Counting down from 3

count = 3
while count > 0:
    print(count)
    count = count - 1
```

## 3. `break` Statement

The `break` statement is used to terminate the loop entirely and transfer control to the statement immediately following the loop. It can be used in both `for` and `while` loops.

```python
# Example with for loop: Stop when 'banana' is found

fruits = ["apple", "banana", "cherry", "date"]
for x in fruits:
  print(x)
  if x == "cherry":
    break
```

```python
# Example with while loop: Stop when i is 3

i = 1
while i < 6:
  print(i)
  if i == 3:
    break
  i = i + 1
```

## 4. `continue` Statement

The `continue` statement is used to skip the rest of the code inside the loop for the current iteration and proceed to the next iteration. It can be used in both `for` and `while` loops.

```python
# Example with for loop: Skip 'banana'

fruits = ["apple", "banana", "cherry"]
for x in fruits:
  if x == "banana":
    continue
  print(x)
```

```python
# Example with while loop: Skip printing 3

i = 0
while i < 5:
  i = i + 1
  if i == 3:
    continue
  print(i)
```

## 5. `enumerate()` Function

- The `enumerate()` function adds a counter to an iterable and returns it as an enumerate object.
- This is particularly useful when you need both the index and the value of items while looping through a list or other sequence.

```python
# Example: Getting index and value from a list

my_list = ["apple", "banana", "cherry"]
for index, value in enumerate(my_list):
  print("Index:", index, "Value:", value)
```

```python
# Example: Starting enumeration from a different number

for index, value in enumerate(my_list, start=1):
  print("Item No.", index, ":", value)
```

## 6. `zip()` Function

- The `zip()` function takes multiple iterable objects (like lists, tuples, etc.) and aggregates elements from each of them into a single iterator of tuples.
- The first element from each iterable is paired, then the second elements, and so on.

```python
# Example: Combining two lists

names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 22]

for name, age in zip(names, ages):
  print(name, "is", age, "years old.")
```

```python
# Example: Combining three lists

countries = ["USA", "Canada", "Mexico"]
codes = [1, 2, 3]
capitals = ["Washington D.C.", "Ottawa", "Mexico City"]

for country, code, capital in zip(countries, codes, capitals):
  print(country, "(", code, ") - Capital:", capital)

# Note: If the iterables have different lengths, zip stops when the shortest iterable is exhausted.
```

### Problem 1: Sum of numbers using a `for` loop

Calculate the sum of all numbers from 1 to 10 (inclusive) using a `for` loop and the `range()` function. Print the final sum.

```python
# Solution 1

total_sum = 0
for num in range(1, 11):
  total_sum = total_sum + num
print("The sum of numbers from 1 to 10 is:", total_sum)
```

### Problem 2: Print even numbers using a `for` loop and `continue`

Iterate through numbers from 1 to 10. Use the `continue` statement to skip odd numbers, printing only the even numbers.

```python
# Solution 2

print("Even numbers from 1 to 10:")
for num in range(1, 11):
  if num % 2 != 0: # Check if the number is odd
    continue
  print(num)
```

### Problem 3: Search for an item using a `for` loop and `break`

You have a list of colors: `['red', 'green', 'blue', 'yellow', 'purple']`. Iterate through the list. If you find the color 'yellow', print 'Yellow found!' and stop the loop immediately using `break`.

```python
# Solution 3

colors = ['red', 'green', 'blue', 'yellow', 'purple']
for color in colors:
  print("Checking:", color)
  if color == 'yellow':
    print("Yellow found!")
    break
```

### Problem 4: Print list items with their position using `enumerate()`

Given a list of cities: `['London', 'Paris', 'Tokyo', 'New York']`. Print each city along with its 1-based position in the list (e.g., '1: London', '2: Paris').

```python
# Solution 4

cities = ['London', 'Paris', 'Tokyo', 'New York']
for position, city in enumerate(cities, start=1):
  print(position, ": ", city)
```

### Problem 5: Combine student names and scores using `zip()`

Given two lists: `student_names = ['Anna', 'Ben', 'Chloe']` and `student_scores = [88, 92, 75]`. Use `zip()` to combine them and print each student's name and their corresponding score.

```python
# Solution 5

student_names = ['Anna', 'Ben', 'Chloe']
student_scores = [88, 92, 75]

for name, score in zip(student_names, student_scores):
  print(name, "scored", score)
```

### Problem 6: Count down using a `while` loop

Write a `while` loop that counts down from 5 to 1, printing each number. After the loop finishes, print 'Countdown complete!'.

```python
# Solution 6

countdown = 5
while countdown >= 1:
  print(countdown)
  countdown = countdown - 1
print("Countdown complete!")
```

### Problem 7: Check for Prime Number

Write a loop to determine if a given number `n = 13` is a prime number. A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself. Print whether the number is prime or not.

```python
# Solution 7

n = 13
is_prime = True

if n < 2:
  is_prime = False
else:
  for i in range(2, n):
    if n % i == 0:
      is_prime = False
      break

if is_prime:
  print(n, "is a prime number.")
else:
  print(n, "is not a prime number.")
```

### Problem 8: Simple Pattern Printing

Use nested `for` loops to print a right-angled triangle pattern of asterisks (`*`) with 5 rows. The first row should have 1 asterisk, the second 2, and so on.

```python
# Solution 8

rows = 5
for i in range(1, rows + 1):
  for j in range(i):
    print("*", end="")
  print()
```

### Problem 9: Check Prime Number with User Input

Write a program that takes an integer as input from the user and determines whether it is a prime number. Print an appropriate message indicating if the number is prime or not.

```python
# Solution 9

num_str = input("Enter an integer: ")

try:
  n = int(num_str)
  is_prime = True

  if n < 2:
    is_prime = False
  else:
    # Only need to check divisors up to the square root of n
    for i in range(2, int(n**0.5) + 1):
      if n % i == 0:
        is_prime = False
        break

  if is_prime:
    print(n, "is a prime number.")
  else:
    print(n, "is not a prime number.")
except ValueError:
  print("Invalid input. Please enter an integer.")
```
