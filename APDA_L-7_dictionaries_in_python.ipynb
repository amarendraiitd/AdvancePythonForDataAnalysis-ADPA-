## Dictionaries

- A dictionary in Python is an unordered collection of data values, used to store data values like a map, which, unlike other data types that hold only a single value as an element, holds a `key:value` pair.
- Each key-value pair in a dictionary maps the key to its associated value.
- Keys must be unique and immutable (e.g., strings, numbers, tuples). Values can be of any data type.

### Creating a Dictionary

```python
# Creating an empty dictionary

empty_dict = {}

empty_dict
```

```python
# Creating a dictionary with integer keys

int_key_dict = {1: 'apple', 2: 'banana', 3: 'cherry'}

int_key_dict
```

```python
# Creating a dictionary with mixed keys (not recommended for good practice, but possible)

mixed_key_dict = {'name': 'Alice', 1: [1, 2, 3]}

mixed_key_dict
```

```python
# Creating a dictionary using the dict() constructor

constructor_dict = dict(a=1, b=2, c=3)

constructor_dict
```

```python
# Creating a dictionary from a list of tuples

tuple_list_dict = dict([('d', 4), ('e', 5)])

tuple_list_dict
```

### Accessing Elements from a Dictionary

```python
my_dict = {'brand': 'Ford', 'model': 'Mustang', 'year': 1964}

# Accessing a value using its key
print("Brand:", my_dict['brand'])
```

```python
# Accessing a value using the get() method
print("Model:", my_dict.get('model'))
```

```python
# Accessing a non-existent key with get() returns None or a default value

print("Color (using get()):", my_dict.get('color'))

print("Color (using get() with default):", my_dict.get('color', 'Not Available'))
```

### Modifying and Adding Elements

```python
my_dict = {'brand': 'Ford', 'model': 'Mustang', 'year': 1964}

# Modifying an existing value
my_dict['year'] = 2020

my_dict
```

```python
# Adding a new key-value pair
my_dict['color'] = 'red'

my_dict
```

### Deleting Elements

```python
my_dict = {'brand': 'Ford', 'model': 'Mustang', 'year': 1964, 'color': 'red'}

# Deleting a specific key-value pair using del keyword

del my_dict['color']

my_dict
```

```python
# Deleting an element using pop() method (returns the value of the removed key)

model = my_dict.pop('model')

print("Removed model:", model)

print("my_dict after popping 'model':", my_dict)
```

```python
# Removing an arbitrary key-value pair using popitem() (removes the last inserted item in Python 3.7+)

item = my_dict.popitem()
print("Removed item:", item)
print("my_dict after popping an item:", my_dict)
```

```python
# Clearing all elements from the dictionary

my_dict = {'a': 1, 'b': 2}
my_dict.clear()

my_dict
```

### Dictionary Methods: `keys()`, `values()`, `items()`

```python
my_dict = {'name': 'Charlie', 'age': 30, 'city': 'New York'}

# Getting all keys
keys = my_dict.keys()

list(keys)
```

```python
# Getting all values
values = my_dict.values()

list(values)
```

```python
# Getting all key-value pairs as tuples
items = my_dict.items()

list(items)
```

### Iterating Through a Dictionary

```python
my_dict = {'fruit': 'apple', 'color': 'green', 'taste': 'sweet'}

# Iterating through keys

for key in my_dict:
    print(key)
```

```python
# Iterating through keys using .keys()

for key in my_dict.keys():
    print(key)
```

```python
# Iterating through values using .values()

for value in my_dict.values():
    print(value)
```

```python
# Iterating through key-value pairs using .items()

for key, value in my_dict.items():
    print(key, value, sep = ":")
```

```python
for key, value in my_dict.items():
    print(key, ":", value)
```

## Dictionary Practice Problems

### Problem 1: Create a Dictionary and Access Elements

Create a dictionary `student` with keys 'name', 'age', 'grade' and values 'Alice', 10, 'A' respectively.
Access and print the 'name' and 'age' of the student.

```python
student = {'name': 'Alice', 'age': 10, 'grade': 'A'}

print("Student Name:", student['name'])
print("Student Age:", student['age'])
```

### Problem 2: Modify and Add Elements to a Dictionary

Given a dictionary `car = {'brand': 'Ford', 'model': 'Mustang', 'year': 1964}`.
Change the 'year' to 2020. Add a new key-value pair 'color': 'red'.
Print the updated dictionary.

```python
car = {'brand': 'Ford', 'model': 'Mustang', 'year': 1964}

car['year'] = 2020
car['color'] = 'red'

print("Updated Car Dictionary:", car)
```

### Problem 3: Get all Keys and Values

Given a dictionary `inventory = {'apples': 50, 'bananas': 20, 'oranges': 30}`.
Print a list of all keys and a list of all values in the dictionary.

```python
inventory = {'apples': 50, 'bananas': 20, 'oranges': 30}

print("Keys:", list(inventory.keys()))
print("Values:", list(inventory.values()))
```

### Problem 4: Check for Key Existence and Delete an Element

Given a dictionary `contacts = {'Alice': '123-456', 'Bob': '789-012'}`.
Check if 'Alice' is a key in the dictionary and print the result.
Then, delete the key 'Bob' and print the dictionary.

```python
contacts = {'Alice': '123-456', 'Bob': '789-012'}

alice_exists = 'Alice' in contacts
print("Is 'Alice' in contacts?", alice_exists)

del contacts['Bob']
print("Contacts after deleting 'Bob':", contacts)
```

### Problem 5: Clear a Dictionary

Create a dictionary `temp_data = {'temp1': 25, 'temp2': 30, 'temp3': 28}`.
Clear all elements from the dictionary. Print the empty dictionary.

```python
temp_data = {'temp1': 25, 'temp2': 30, 'temp3': 28}

temp_data.clear()
print("Cleared dictionary:", temp_data)
```
