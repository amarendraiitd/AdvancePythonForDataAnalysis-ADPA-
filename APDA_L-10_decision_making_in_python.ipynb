# Decision Making in Python (if, elif, else)

## Introduction to Decision Making

In programming, decision making allows your code to perform different actions based on different conditions. This is fundamental for creating programs that can respond intelligently to various situations. Python provides several statements for decision making: `if`, `if-else`, and `if-elif-else`.

## The `if` Statement

The `if` statement is the simplest form of decision making. It executes a block of code only if a specified condition is true.

### Syntax:
```python
if condition:
    # code to execute if condition is true
```

- The `condition` is an expression that evaluates to either `True` or `False`.
- The code block indented below the `if` statement is executed only when the condition is `True`.

### Example: `if` Statement

````python
score = 75

if score >= 60:
    print("You passed the exam!")

print("End of program.")
````

## The `if-else` Statement

The `if-else` statement allows you to execute one block of code if the condition is true and a different block of code if the condition is false.

### Syntax:
```python
if condition:
    # code to execute if condition is true
else:
    # code to execute if condition is false
```

- If the `condition` is `True`, the code block under `if` is executed.
- If the `condition` is `False`, the code block under `else` is executed.

### Example: `if-else` Statement

````python
age = 17

if age >= 18:
    print("You are old enough to vote.")
else:
    print("You are not yet old enough to vote.")

print("Voting eligibility check complete.")
````

## The `if-elif-else` Statement

The `if-elif-else` statement is used when you have multiple conditions to check. It allows you to test several conditions and execute a different block of code for the first condition that evaluates to true.

### Syntax:
```python
if condition1:
    # code to execute if condition1 is true
elif condition2:
    # code to execute if condition2 is true
elif condition3:
    # code to execute if condition3 is true
else:
    # code to execute if none of the above conditions are true
```

- Conditions are checked sequentially.
- The code block for the *first* `True` condition is executed.
- If all `if` and `elif` conditions are `False`, the `else` block (if present) is executed.

### Example: `if-elif-else` Statement

````python
temperature = 25

if temperature > 30:
    print("It's a hot day!")
elif temperature > 20:
    print("It's a pleasant day.")
elif temperature > 10:
    print("It's a cool day.")
else:
    print("It's a cold day!")

print("Weather update complete.")
````

## Nested `if` Statements

You can place an `if` statement inside another `if`, `elif`, or `else` statement. This is known as nesting, and it allows for more complex decision-making logic.

### Syntax:
```python
if outer_condition:
    # code for outer_condition true
    if inner_condition:
        # code for inner_condition true
    else:
        # code for inner_condition false
else:
    # code for outer_condition false
```

- The inner `if` statement is only evaluated if the outer `if` condition is `True`.

### Example: Nested `if` Statements

````python
has_ticket = True
owns_popcorn = True

if has_ticket:
    print("Welcome to the cinema!")
    if owns_popcorn:
        print("Enjoy your movie with popcorn!")
    else:
        print("You might want to grab some popcorn from the concession stand.")
else:
    print("Please purchase a ticket to enter.")

print("Cinema entry process complete.")
````

### Problem 1: Positive, Negative, or Zero

Write a program that takes a number and prints whether it's positive, negative, or zero.

````python
# Solution 1
number = -5

if number > 0:
    print("Positive")
elif number < 0:
    print("Negative")
else:
    print("Zero")
````

### Problem 2: Check Even or Odd

Write a program that checks if a given integer is even or odd.

````python
# Solution 2
num = 4

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
````

### Problem 3: Eligibility for Discount

A store offers a 10% discount if a customer spends more than $100. If they spend exactly $100, they get a 5% discount. Otherwise, there is no discount.

Write a program to calculate the final amount after applying the discount, given the `purchase_amount`.

````python
# Solution 3
purchase_amount = 120

if purchase_amount > 100:
    final_amount = purchase_amount * 0.90  # 10% discount
elif purchase_amount == 100:
    final_amount = purchase_amount * 0.95  # 5% discount
else:
    final_amount = purchase_amount

print("Original purchase amount:", purchase_amount)
print("Final amount after discount:", final_amount)
````

### Problem 4: Grade Calculator

Write a program that takes a student's score (out of 100) and assigns a grade based on the following criteria:
- 90-100: A
- 80-89: B
- 70-79: C
- 60-69: D
- Below 60: F

````python
# Solution 4
score = 85

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
elif score >= 60:
    grade = 'D'
else:
    grade = 'F'

print("Score:", score)
print("Grade:", grade)
````

### Problem 5: Login Check (Nested `if`)

Create a program that checks if a user is logged in. If they are logged in, then check if they are an administrator. Print appropriate messages.

````python
# Solution 5
is_logged_in = True
is_admin = False

if is_logged_in:
    print("User is logged in.")
    if is_admin:
        print("Welcome, Administrator!")
    else:
        print("Welcome, Regular User.")
else:
    print("User is not logged in. Please log in to continue.")
````
