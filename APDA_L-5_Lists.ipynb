    - List concatenation: to add/join two lists

```python
my_list = [11, 24, 2, 9, 2, 10, 22.5]
my_list2 = [100, 50, "Hello"]

# method1: using + operator

conc_list1 = my_list + my_list2
print(conc_list1)
```

```python
my_list = [11, 24, 2, 9, 2, 10, 22.5]
my_list2 = [100, 50, "Hello"]

# method2: using extend() method

my_list.extend(my_list2)
print(my_list)
```

```python
my_list = [11, 24, 2, 9, 2, 10, 22.5]
my_list2 = [100, 50, "Hello"]

# method3: using append() method in loop

for item in my_list2:
    my_list.append(item)

print(my_list)
```

```python
my_list = [11, 24, 2, 9, 2, 10, 22.5]
my_list2 = [100, 50, "Hello"]

# method4: using unpacking operator

conc_list3 = [*my_list, *my_list2]

print(conc_list3)
```

    Copy a List:
- used if you want to perform operations on list without disturbing original list.
- you can not copy a list by list2=list1, because list2 will only be a reference to list1. changes in list1 will automatically also be made in list2.

        - list.copy():Return a shallow copy of the list.

        - method2: list() method   

```python
my_list = [11, 24, 2, 9, 2, 10, 22.5]

my_list_copy = my_list.copy()
my_list.pop()

print(my_list)
print(my_list_copy)
```

```python
# method2: list() method

my_list = [11, 24, 2, 9, 2, 10, 22.5]
my_list_copy = list(my_list)

print(my_list_copy)
```

    -List comprehension:
    

```python
# list comprehension:

my_list = [11, 24, 2, 9, 2, 10, 22.5]
squared_list = [item**2 for item in my_list]
print(squared_list)
```

```python
# shorter way to do the same thing:
row1 = [255, 0, 128]
row1 = row1 + row2

# which method is better? extend() or + operator?
# extend() method is generally more efficient than using the + operator for concatenating lists, especially when dealing with large lists. 
# The + operator creates a new list and copies the elements from both lists, which can be slower and consume more memory. 
# On the other hand, the extend() method modifies the original list in place without creating a new one, making it more efficient for larger lists.
```
