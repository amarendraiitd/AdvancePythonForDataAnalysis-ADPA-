## Introduction to Tuples

- A tuple is an ordered, immutable collection of items in Python.
- Tuples are similar to lists, but unlike lists, their contents cannot be changed after creation.
- This immutability makes them useful for representing fixed collections of items, and they can be used as dictionary keys.

- difference between list and tuple



```python
# Example: A simple tuple

my_tuple = (1, 2, 'hello', 3.14)

print(my_tuple)
print(type(my_tuple))
```

## Creating Tuples

Tuples can be created in several ways:

*   **Empty Tuple**: Using empty parentheses `()`.
*   **Tuple with one element**: Needs a trailing comma `(item,)`.
*   **Tuple with multiple elements**: Separating items with commas inside parentheses `(item1, item2, ...)`.
*   **Tuple Packing**: Assigning multiple values to a single variable, automatically creating a tuple.
*   **Tuple Unpacking**: Assigning elements of a tuple to multiple variables.

```python
# 1. Empty tuple
empty_tuple = ()
print(empty_tuple)

# 2. Tuple with one element (note the comma)
single_element_tuple = (42,)
print(single_element_tuple)

# Without the comma, it's just an int
not_a_tuple = (42)
print(not_a_tuple)

# 3. Tuple with multiple elements
multiple_elements_tuple = (10, 20, 30)
print(multiple_elements_tuple)

# 4. Tuple packing
packed_tuple = 1, 'apple', True
print(packed_tuple)

# 5. Tuple unpacking
a, b, c = packed_tuple
print(a,b,c)
```

## Accessing Elements in Tuples

Elements in a tuple can be accessed using indexing and slicing, similar to lists and strings.

*   **Indexing**: Access individual elements using square brackets `[]` with an integer index (0-based).
*   **Slicing**: Extract a sub-tuple (a portion of the tuple) using the slice operator `[:]`.

```python
my_data = ('apple', 'banana', 'cherry', 'date', 'elderberry')

# Accessing elements by index
print(f"First element: {my_data[0]}")
print(f"Third element: {my_data[2]}")
print(f"Last element (negative indexing): {my_data[-1]}")

# Slicing tuples
print(f"Elements from index 1 to 3 (exclusive): {my_data[1:4]}")
print(f"Elements from the beginning to index 2 (exclusive): {my_data[:2]}")
print(f"Elements from index 3 to the end: {my_data[3:]}")
print(f"All elements (a copy): {my_data[:]}")
```

```python
# we can create sub tuples from orginal tuple using slicing

tuple1 = (2, 4, "Hello", True, None)

tuple2 = tuple1[1:]
print(tuple2)
```

## Advanced Tuple Examples and Use Cases

Tuples are versatile and appear in many Python contexts. Here are a few examples showcasing their utility beyond simple data storage:

*   **Returning Multiple Values from Functions**: Functions can return multiple values, and Python automatically packs these into a tuple.
*   **Using Tuples as Dictionary Keys**: Due to their immutability, tuples can serve as keys in dictionaries, which lists cannot.
*   **Nested Tuples**: Tuples can contain other tuples, allowing for complex data structures.
*   **Converting between Lists and Tuples**: Sometimes you need the mutability of a list and sometimes the immutability of a tuple. Python provides easy ways to convert between them.

```python
# Example 1: Returning multiple values from a function

def get_coordinates():
    return 10, 20  # Python automatically packs these into a tuple

x, y = get_coordinates()
print(x,y)

coord_tuple = get_coordinates()
print(coord_tuple)
```

```python
# Example 2: Using tuples as dictionary keys
# Tuples are hashable (immutable), so they can be dictionary keys.
# Lists are not hashable (mutable), so they cannot.

location_data = {
    (40.7128, -74.0060): "New York City",
    (34.0522, -118.2437): "Los Angeles"
}

print(f"City at (40.7128, -74.0060): {location_data[(40.7128, -74.0060)]}")
```

```python
# Example 3: Nested Tuples

nested_tuple = (1, (2, 3), ('a', 'b', ('c', 4)))

print(nested_tuple)

print(nested_tuple[2][2][0])
```

```python
# Example 4: Converting between lists and tuples

my_list = [1, 2, 3]
converted_tuple = tuple(my_list)
print(f"List converted to tuple: {converted_tuple}")

my_tuple_to_convert = ('x', 'y', 'z')
converted_list = list(my_tuple_to_convert)
print(f"Tuple converted to list: {converted_list}")
```

## Basic Tuple Operations

Even though tuples are immutable, you can perform several operations that result in new tuples.

*   **Concatenation**: Combine two or more tuples using the `+` operator.
*   **Repetition**: Repeat the elements of a tuple multiple times using the `*` operator.
*   **Length**: Get the number of elements in a tuple using the `len()` function.
*   **Membership**: Check if an element exists in a tuple using the `in` and `not in` operators.

```python
# Concatenation

tuple1 = (1, 2, 3)
tuple2 = ('a', 'b', 'c')
combined_tuple = tuple1 + tuple2

combined_tuple
```

```python
# Repetition

repeated_tuple = tuple1 * 3

repeated_tuple
```

```python
# Length
len(combined_tuple)
```

```python
# Membership

tuple1 = (1, 2, 3)
tuple2 = ('a', 'b', 'c')

print(5 in tuple1)

print('x' not in tuple2)
```

## Tuple Methods

Tuples have a limited number of built-in methods due to their immutability:

*   `tuple.count(element)`: Returns the number of times a specified element occurs in a tuple.
*   `tuple.index(element)`: Searches the tuple for a specified element and returns its position (index) of the first occurrence. Raises a `ValueError` if the element is not found.
*   `tuple.index(element, start, end)`: 
    - element: The item you want to search for.
    - start (Optional): The index to start the search from. 
    - end (Optional): The index to stop the search before.

```python
my_tuple = (1, 2, 2, 3, 4, 2, 5)

# count() method
count_of_2 = my_tuple.count(2)
print(count_of_2)

# index() method
index_of_3 = my_tuple.index(2)
print(index_of_3)
```

```python
numbers = (1, 2, 3, 4, 2, 5, 2)
# Search for the number 2 starting at index 3
print(numbers.index(2, 3))
```

```python
tuple1 = (2, 4, 2, 2, "Hello", True, None)

tuple1.index(2, 2, 4) # value, start, end
```

```python
try:
    my_tuple.index(99)
except ValueError as e:
    print(e)
```

## Immutability of Tuples

- A key characteristic of tuples is that they are **immutable**. (similar to string)
- This means that once a tuple is created, you cannot change its elements (add, remove, or modify).
- If you need a collection that can be changed, use a list instead.

- Attempts to modify a tuple will result in a `TypeError`.

```python
my_immutable_tuple = (10, 20, 30)

# Attempting to change an element will raise a TypeError
try:
    my_immutable_tuple[0] = 5
except TypeError as e:
    print(e)

# Attempting to add an element
try:
    my_immutable_tuple.append(40) # Tuples do not have an append method
except AttributeError as e:
    print(e)

# You can, however, reassign the variable to a new tuple
my_immutable_tuple = (5, 20, 30)
print(my_immutable_tuple)
```

    - to make tuple mutable:
    - convert tuple to list.
    - then you can apply all list methods like append(), pop(), insert(), index(), etc.
    - you can then convert back list to tuple

```python
tuple1 = (2, 4, "Hello", True, None)

list_tuple = list(tuple1)

list_tuple.append("Python")
print(list_tuple)

tuple1 = tuple(list_tuple)
print(tuple1)
```

```python
len(tuple1)
```

### Problem 1: Swap Variables

Write a Python code snippet to swap the values of two variables `a` and `b` without using a temporary variable, leveraging tuple packing and unpacking.

```python
a = 5
b = 10

print("Before swap:")
print("a:", a)
print("b:", b)

a, b = b, a # Tuple packing and unpacking for swapping

print("\nAfter swap:")
print("a:", a)
print("b:", b)
```

### Problem 2: Function Returning Min/Max

Create a function that takes a list of numbers and returns a tuple containing the minimum and maximum values from that list.

```python
def get_min_max(numbers):
    if not numbers:
        return None
    return (min(numbers), max(numbers))

my_numbers = [10, 3, 20, 7, 15]

result = get_min_max(my_numbers)

print("Original list:", my_numbers)
print("Min and Max:", result)
```

### Problem 3: Count Vowels

Given a string, create a tuple of vowels ('a', 'e', 'i', 'o', 'u'). Then, write code to count how many times each vowel appears in the string. Store the counts in a dictionary where keys are vowels and values are their counts.

```python
# Solution 3

text = "Hello, World! This is a test string for vowel counting."

vowels = ('a', 'e', 'i', 'o', 'u')

vowel_counts = {}

for char in text.lower():
    if char in vowels:
        vowel_counts[char] = vowel_counts.get(char, 0) + 1

print("Text:", text)
print("Vowel counts:", vowel_counts)
```

### Problem 4: Tuple of Records

Imagine you have student data (name, age, grade). Create a list of tuples, where each tuple represents a student's record. Then, write code to iterate through this list and print each student's name and age.

```python
# Solution 4

student_records = [
    ("Alice", 18, "A"),
    ("Bob", 19, "B+"),
    ("Charlie", 17, "A-")
]

print("Student Records:")
for record in student_records:
    name, age, _ = record # Unpacking the tuple, ignoring grade
    print("Name:", name, ", Age:", age)
```
