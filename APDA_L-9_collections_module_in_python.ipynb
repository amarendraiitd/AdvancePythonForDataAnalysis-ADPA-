## Python's `collections` Module

Python's `collections` module implements specialized container datatypes providing alternatives to Python’s general purpose built-in containers like `dict`, `list`, `set`, and `tuple`. These specialized containers offer additional functionality or performance characteristics for specific use cases.

We will explore some of the most commonly used types from this module.

### `collections.Counter`

A `Counter` is a `dict` subclass for counting hashable objects. It is an unordered collection where elements are stored as dictionary keys and their counts are stored as dictionary values. Counts are allowed to be any integer value, including zero or negative counts.

#### Characteristics of `Counter`:

1.  **Counting**: Easily counts occurrences of items in an iterable.
2.  **Dictionary-like**: Behaves much like a regular dictionary, supporting all `dict` methods.
3.  **Convenient methods**: Provides methods like `elements()` for iterating over elements (repeating each as many times as its count) and `most_common(n)` for listing the 'n' most frequent elements.
4.  **Arithmetic operations**: Supports addition, subtraction, intersection, and union with other `Counter` objects.

#### Creating and Using `Counter`

```python
from collections import Counter

# Creating a Counter from a list
my_list = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']
fruit_counts = Counter(my_list)
print("Fruit counts:", fruit_counts)

# Creating a Counter from a string
word = "hello world"
char_counts = Counter(word)
print("Character counts:", char_counts)

# Creating an empty Counter and updating it
empty_counter = Counter()
empty_counter.update(['cat', 'dog', 'cat'])
print("Updated counter:", empty_counter)

# Accessing counts
print("Count of 'apple':", fruit_counts['apple'])
print("Count of 'grape' (non-existent):", fruit_counts['grape']) # Returns 0 for non-existent items

# Elements method
print("Elements of fruit_counts:", list(fruit_counts.elements()))

# Most common elements
print("Two most common fruits:", fruit_counts.most_common(2))

# Arithmetic operations (addition)
c1 = Counter(['a', 'b', 'b'])
c2 = Counter(['b', 'c', 'c'])
print("c1 + c2:", c1 + c2)

# Arithmetic operations (subtraction)
print("c1 - c2:", c1 - c2)
```

### `collections.defaultdict`

A `defaultdict` is a subclass of the built-in `dict` class. It overrides one method and adds one writable instance variable. The main feature of `defaultdict` is that it allows you to specify a default value factory (a function that provides a default value) for missing keys. When you try to access a key that doesn't exist, instead of raising a `KeyError`, `defaultdict` automatically creates an entry for that key with the default value provided by the factory function.

#### Characteristics of `defaultdict`:

1.  **Handles missing keys**: Automatically inserts a default value when a non-existent key is accessed, avoiding `KeyError`.
2.  **Factory function**: Requires a factory function (e.g., `list`, `int`, `set`, `str`) as an argument during initialization.
3.  **Dictionary-like**: Supports all standard dictionary operations.

#### Creating and Using `defaultdict`

```python
from collections import defaultdict

# Example 1: Grouping items using a list as default factory
# If a key is accessed for the first time, an empty list is created.
my_list_of_tuples = [('animal', 'dog'), ('vehicle', 'car'), ('animal', 'cat'), ('fruit', 'apple'), ('vehicle', 'bike')]

grouped_items = defaultdict(list)
for key, value in my_list_of_tuples:
    grouped_items[key].append(value)

print("Grouped items (list factory):", grouped_items)
print("Accessing a new key 'color':", grouped_items['color']) # Creates an empty list
print("Grouped items after accessing new key:", grouped_items)


# Example 2: Counting occurrences using int as default factory
# If a key is accessed for the first time, 0 is created.
word = 'banana'
char_counts = defaultdict(int)
for char in word:
    char_counts[char] += 1

print("Character counts (int factory):", char_counts)
print("Accessing a new key 'z':", char_counts['z']) # Creates 0
print("Character counts after accessing new key:", char_counts)


# Example 3: Storing unique items using set as default factory
# If a key is accessed for the first time, an empty set is created.
relationships = [('Alice', 'Bob'), ('Alice', 'Charlie'), ('Bob', 'David'), ('Alice', 'Bob')]
connections = defaultdict(set)
for person1, person2 in relationships:
    connections[person1].add(person2)

print("Connections (set factory):", connections)
print("Accessing a new key 'Eve':", connections['Eve']) # Creates an empty set
print("Connections after accessing new key:", connections)
```

### `collections.deque`

A `deque` (double-ended queue) is a list-like container with fast appends and pops on either end. It is optimized for adding and removing items from both the beginning and the end of the sequence. While lists provide similar operations, they become inefficient for large collections when operations are performed at the beginning (e.g., `insert(0, item)` or `pop(0)`), as all other elements need to be shifted.

#### Characteristics of `deque`:

1.  **Efficient Appends/Pops**: Constant-time (`O(1)`) performance for adding and removing elements from either end.
2.  **Mutable**: Elements can be added, removed, and modified.
3.  **Iterable**: Can be iterated over like a list.
4.  **Fixed-size (optional)**: Can be initialized with a `maxlen` to create a fixed-size queue, where new items automatically push out the oldest items.

#### Creating and Using `deque`

```python
from collections import deque

# Creating a deque
my_deque = deque([1, 2, 3, 4, 5])
print("Initial deque:", my_deque)

# Appending elements to the right
my_deque.append(6)
print("After append(6):", my_deque)

# Appending elements to the left
my_deque.appendleft(0)
print("After appendleft(0):", my_deque)

# Popping elements from the right
popped_right = my_deque.pop()
print("Popped from right:", popped_right, ", Deque now:", my_deque)

# Popping elements from the left
popped_left = my_deque.popleft()
print("Popped from left:", popped_left, ", Deque now:", my_deque)

# Extending with an iterable
my_deque.extend([6, 7])
print("After extend([6, 7]):", my_deque)

# Extending left with an iterable
my_deque.extendleft([-1, -2])
print("After extendleft([-1, -2]):", my_deque)

# Rotating the deque
my_deque.rotate(2) # Rotate 2 steps to the right
print("After rotate(2):", my_deque)
my_deque.rotate(-1) # Rotate 1 step to the left
print("After rotate(-1):", my_deque)

# Creating a fixed-size deque (maxlen)
fixed_deque = deque(maxlen=3)
fixed_deque.append(1)
fixed_deque.append(2)
fixed_deque.append(3)
print("Fixed deque (3 elements):", fixed_deque)
fixed_deque.append(4) # Appends 4, pushes out 1
print("Fixed deque after append(4):", fixed_deque)
fixed_deque.appendleft(0) # Appends 0, pushes out 2
print("Fixed deque after appendleft(0):", fixed_deque)
```

### `collections.namedtuple`

`namedtuple` is a factory function for creating tuple subclasses with named fields. This means you can create lightweight, immutable object types where elements can be accessed using names instead of just integer indices. It's often used when you need to group a few related pieces of data together, but don't need the full functionality of a class.

#### Characteristics of `namedtuple`:

1.  **Readability**: Access fields by name (e.g., `point.x`) instead of index (e.g., `point[0]`), making code more understandable.
2.  **Immutability**: Like regular tuples, `namedtuple` instances are immutable once created.
3.  **Lightweight**: More memory-efficient than a custom class for simple data structures.
4.  **Interoperability**: Can be used wherever regular tuples are expected.

#### Creating and Using `namedtuple`

```python
from collections import namedtuple

# Define a namedtuple type for a Point (x, y coordinates)
# The first argument is the name of the tuple type, the second is a string of field names separated by spaces or commas.
Point = namedtuple('Point', 'x y')

# Create instances of the Point namedtuple
p1 = Point(10, 20)
p2 = Point(y=30, x=40) # Fields can be assigned by name

print("Point 1:", p1)
print("Point 2:", p2)

# Accessing elements by name
print("p1.x:", p1.x)
print("p2.y:", p2.y)

# Accessing elements by index (like a regular tuple)
print("p1[0]:", p1[0])

# namedtuple is immutable, so trying to change a field will raise an error
# Uncommenting the line below would cause an error:
# p1.x = 15

# Convert namedtuple to a dictionary (useful for serialization or inspection)
print("p1 as dictionary:", p1._asdict())

# Replacing fields (creates a new namedtuple instance)
p3 = p1._replace(x=100)
print("p3 (p1 with x replaced):", p3)

# Namedtuple for a Car (make, model, year)
Car = namedtuple('Car', ['make', 'model', 'year'])
my_car = Car('Toyota', 'Camry', 2020)
print("My car:", my_car)
print("Car model:", my_car.model)
```

### Problem 4: Counter Usage

Use `collections.Counter` to count the occurrences of items in the list `items = ['red', 'blue', 'red', 'green', 'blue', 'red']`. Print the resulting `Counter` object and then print the count of the item `'red'`.

### Problem 5: Deque Operations

Create a `collections.deque` with initial elements `[1, 2, 3]`. Add `0` to the left end and `4` to the right end. Then, remove an element from the left end and print the final state of the deque.

```python
# Solution 5
from collections import deque

my_deque = deque([1, 2, 3])
my_deque.appendleft(0)
my_deque.append(4)
my_deque.popleft()

print("Final deque:", my_deque)
```

### Problem 6: Namedtuple Creation and Access

Define a `collections.namedtuple` called `Book` with fields `title` and `author`. Create an instance for the book titled "The Hitchhiker's Guide to the Galaxy" by "Douglas Adams". Print the book's title using its field name.

```python
# Solution 6
from collections import namedtuple

Book = namedtuple('Book', 'title author')
my_book = Book('The Hitchhiker\'s Guide to the Galaxy', 'Douglas Adams')

print("Book title:", my_book.title)
```
